```python
import random
def quicksort(L):
  if len(L) < 2:
    return L
  else:
    pivot = L[len(L) // 2]
    less, equal, larger = [], [], []

    for item in L:
      if item < pivot:
        less.append(item)
      elif item == pivot:
        equal.append(item)
      else:
        larger.append(item)
    return quicksort(less) + equal + quicksort(larger)

quicksort([1,2,5,1,6,9,32,4,16,7,1])

```

```python
class Car:
  def __init__(self, make, model, year, car_type, engine, acceleration):
    self.make = make
    self.model = model
    self.year = year
    self.car_type = car_type
    self.engine = engine
    self.num_wheels = 5
    self.num_doors = 4
    self.engine_running = False
    self.speed = 0
    self.acceleration = acceleration

  def engine_start(self):
    self.engine_running = True
    print("Vroom! Engine Started")

  def engine_stop(self):
    if not self.engine_running:
      print("Engine is already off")
    else:
      self.engine_running = False
      print("Engine Stopped")
  def acceleration(self, new_speed):
    self.speed = new_speed
    if self.speed > 150:
      print("You are driving too fast")
      self.speed -= new_speed
  def __str__(self):
    return f"Make: {self.make}; Model: {self.model}; Year: {self.year}"
```

```python
class C:
  def __init__(self, a):
    self.a = a
    self.b = [1, 2, 3, 4]

  def __add__(self, other):
    return self.a + other.a

  def __mul__(self, other):
    return self.a * other.a

  def __eq__(self, other):
    return self.a == other.a

  def __len__(self):
    return len(self.b)

  def __getitem__(self, idx):
    return self.b[idx]

  def __setitem__(self, idx, item):
    self.b[idx] = item

x = C(3)
y = C(5)
```

```python
class ATM:
  def __init__(self, balance):
    self.__balance = balance
    self.__employed = False

  def get_employment_status(self):
    return self.__employed

  def set_employment_status(self, newstatus):
    assert type(newstatus) == bool, "Employment status must be a boolean value"
    self.__employed = newstatus

  def deposit(self, amount):
    self.__balance += amount
    print(f"New account balance: ${self.__balance}")

  def withdral(self, amount):
    if self.__balance < amount :
      print(f"Insufficient funds...")
    else:
      self.__balance -= amount
      print(f"New account balance: ${self.__balance}")
```

```python
account = ATM(1000)
account.deposit(200)
```

```python
account.withdral(500)
account.withdral(800)
```

```python
class Bird:
  def __init__(self):
    print("Bird is created")
  def whoAmI(self):
    print("I am Bird")
  def fly(self):
    print("Birds can fly")
  def swim(self):
    print("Birds can't swim")

class Owl(Bird):
  def __init__(self):
    super().__init__()
    print("Owl is created")
  def night_vision(self):
    print("Owls have night vision")
  def whoAmI(self):
    print("I am an Owl")

class Penguin(Bird):
  def __init__(self):
    super().__init__()
    print("Penguin is created")
  def swim(self):
    print("Penguins can swim")
  def whoAmI(self):
    print("I am a Penguin")
```
